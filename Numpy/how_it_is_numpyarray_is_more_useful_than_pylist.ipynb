```python
import numpy as np
import time
```

```python
size = 1_000_000
#python list 
py_list = list(range(size))
start = time.time()
#lets square the whole list
sq_list = [x**2 for x in py_list]
end = time.time()

print(f"python list time is= {end-start} seconds")
```

```python
size = 1_000_000
#python list 
np_arr = np.array(range(size))
start = time.time()
#lets square the whole list
sq_array = np_arr**2
end = time.time()

print(f"numpy array time is= {end-start} seconds")
```

```python
python list time is= 0.12420225143432617 seconds
numpy array time is= 0.002803802490234375 seconds
```

```python
numpy stores data in contiguous memory blocks
uses less memory
builtin mathematical functions
supports direct vectorized operations
```
